## importing dependencies

```python
import pandas as pd
import glob
import pyreadstat
```

## Define the path to the CSV files / create ihme dataframe

```python
csv_files_path = "../data/ihme_data/*.csv"
```

## Get a list of all CSV file paths

```python
csv_files = glob.glob(csv_files_path)
```

### Read each CSV file and append it to the list

```python
dataframes = []

for file in csv_files: 
    df = pd.read_csv(file) 
    dataframes.append(df)
```

### Merge all DataFrames into a single DataFrame

```python
ihme = pd.concat(dataframes, ignore_index=True)
```

### Save the merged DataFrame to a new CSV file

```python
ihme.to_csv("../data/ihme.csv", index=False)
ihme.head()
```

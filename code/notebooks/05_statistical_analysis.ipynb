```python
# Mapping SDI regions to numeric rank
sdi_rank = {
    'Low SDI': 1,
    'Low-middle SDI': 2,
    'Middle SDI': 3,
    'High-middle SDI': 4,
    'High SDI': 5
}
# Filter only age-standardized, both sexes
subset = qci[
    (qci['age_name'] == 'Age-standardized') &
    (qci['sex_name'] == 'Both') &
    (qci['iso_location_name'].isin(sdi_rank.keys()))
].copy()
# Add SDI rank
subset['sdi_score'] = subset['iso_location_name'].map(sdi_rank)
# Drop missing
subset = subset.dropna(subset=['sdi_score', 'qci'])
```

```python
subset
```

```python
# Compute correlation per year
correlation_by_year = []

for year in sorted(subset['year'].unique()):
    df_year = subset[subset['year'] == year].dropna(subset=['sdi_score', 'pca_score'])
    if len(df_year) >= 3:  # Need at least 3 points for correlation
        rho, p = spearmanr(df_year['sdi_score'], df_year['pca_score'])
        correlation_by_year.append({'year': year, 'rho': rho, 'p_value': p})

correlation_df = pd.DataFrame(correlation_by_year)
print(correlation_df)
```

```python
qci_merged_sdi = qci.merge(
    sdi_filtered[['sdi_location_name', 'year', 'sdi_value']],
    on=['sdi_location_name', 'year'],
    how='left'
)
```

```python
qci_merged_sdi.head(5)
```

```python

# Filter only age-standardized, both sexes
filtered_sdi_1990_2021 = qci_merged_sdi[
    (qci_merged_sdi['age_name'] == 'Age-standardized') &
    (qci_merged_sdi['sex_name'] == 'Both')
].copy()

# Compute correlation per year
correlation_by_year = []

for year in sorted(filtered_sdi_1990_2021['year'].unique()):
    df_year = filtered_sdi_1990_2021[filtered_sdi_1990_2021['year'] == year].dropna(subset=['sdi_value', 'pca_score'])
    if len(df_year) >= 3:  # Need at least 3 points for correlation
        rho, p = spearmanr(df_year['sdi_value'], df_year['pca_score'])
        correlation_by_year.append({'year': year, 'rho': rho, 'p_value': p})

correlation_sdi_1990_2021_df = pd.DataFrame(correlation_by_year)
print(correlation_sdi_1990_2021_df)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.plot(correlation_sdi_1990_2021_df['year'], correlation_sdi_1990_2021_df['rho'], marker='o')
plt.title('Spearman Correlation between SDI and QCI (1990–2021)')
plt.xlabel('Year')
plt.ylabel('Spearman Rho')
plt.grid(True)
plt.ylim(0.6, 0.8)
plt.tight_layout()
plt.show()
```

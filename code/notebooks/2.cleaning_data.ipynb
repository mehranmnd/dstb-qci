## removing unwanted rows

```python
wanted_years =[1990,1991,1992,1993,1994,1995,1996,1997,1998,1999,2000,
                 2001,2002,2003,2004,2005,2006,2007,2008,2009,2010,2011,
                 2012,2013,2014,2015,2016,2017,2018,2019,2020,2021]
ihme = ihme[ihme["year"].isin(wanted_years)]
print(ihme.year.unique())
```

### Exclude rows where location_id equals 533 (georgia state)

```python
print(ihme.shape)
ihme = ihme[ihme['location_id'] != 533]
print(ihme.shape)
```

### remove rows where metric name is percent

```python
print(ihme.shape)
condition1 = (ihme['metric_name'] == "Percent")
ihme = ihme[~condition1]  # Invert the condition using ~
print(ihme.shape)
```

## remove rows where location_name are null

```python
print(ihme.info(show_counts=True))
```

```python
ihme = ihme[ihme['location_name'].notna()]
print(ihme.info(show_counts=True))
```

## simplify measure names

```python
ihme.loc[ihme['measure_name'] == "DALYs (Disability-Adjusted Life Years)", 'measure_name'] = "DALYs"
ihme.loc[ihme['measure_name'] == "YLDs (Years Lived with Disability)", 'measure_name'] = "YLDs"
ihme.loc[ihme['measure_name'] == "YLLs (Years of Life Lost)", 'measure_name'] = "YLLs"
```

# preparation for model training

## Filter rows where metric_name is "Rate"

```python
main_data_long_Rate = ihme[ihme['metric_name'] == "Rate"]
print(main_data_long_Rate['age_name'].unique())
print(main_data_long_Rate.shape)
```

## Reshape main_data_long_Rate

```python
main_data_wide_Rate = main_data_long_Rate.pivot_table(
    index=["iso_location_name", "location_name",'haq_location_name', "year", "sex_name", "age_name"],
    columns="measure_name",
    values=["val", "upper", "lower"]
).reset_index()
main_data_wide_Rate.head(2)
```

## Flatten the column hierarchy created by pivot_table

```python
main_data_wide_Rate.columns = ['_'.join(filter(None, col)).rstrip('_') for col in main_data_wide_Rate.columns.values]
```

```python
main_data_wide_Rate.head(2)
```

## calculations

### MIR: Mortality to Incidence Ratio

```python
main_data_wide_Rate['MIR'] = main_data_wide_Rate['val_Deaths'] / main_data_wide_Rate['val_Incidence']
main_data_wide_Rate['lower_MIR'] = main_data_wide_Rate['lower_Deaths'] / main_data_wide_Rate['lower_Incidence']
main_data_wide_Rate['upper_MIR'] = main_data_wide_Rate['upper_Deaths'] / main_data_wide_Rate['upper_Incidence']
```

### YLLs to YLDs Ratio

```python
main_data_wide_Rate['YLLtoYLD'] = main_data_wide_Rate['val_YLLs'] / main_data_wide_Rate['val_YLDs']
main_data_wide_Rate['lower_YLLtoYLD'] = main_data_wide_Rate['lower_YLLs'] / main_data_wide_Rate['lower_YLDs']
main_data_wide_Rate['upper_YLLtoYLD'] = main_data_wide_Rate['upper_YLLs'] / main_data_wide_Rate['upper_YLDs']
```

### DALY to Prevalence Ratio

```python
main_data_wide_Rate['DALtoPER'] = main_data_wide_Rate['val_DALYs'] / main_data_wide_Rate['val_Prevalence']
main_data_wide_Rate['lower_DALtoPER'] = main_data_wide_Rate['lower_DALYs'] / main_data_wide_Rate['lower_Prevalence']
main_data_wide_Rate['upper_DALtoPER'] = main_data_wide_Rate['upper_DALYs'] / main_data_wide_Rate['upper_Prevalence']
```

### Prevalence to Incidence Ratio

```python
main_data_wide_Rate['PERtoINC'] = main_data_wide_Rate['val_Prevalence'] / main_data_wide_Rate['val_Incidence']
main_data_wide_Rate['lower_PERtoINC'] = main_data_wide_Rate['lower_Prevalence'] / main_data_wide_Rate['lower_Incidence']
main_data_wide_Rate['upper_PERtoINC'] = main_data_wide_Rate['upper_Prevalence'] / main_data_wide_Rate['upper_Incidence']
```

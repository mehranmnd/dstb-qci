```python
import pandas as pd
import plotly.express as px
import matplotlib.pyplot as plt
```

```python
# Load the data
df = pd.read_csv("../results/qci.csv")
df['year'] = df['year'].astype(int)
```

```python
# Filter for age-standardized and sex = Both
standardized_heatmap_data = df[df['sex_name'] == 'Both']
```

```python
fig = px.choropleth(
    standardized_heatmap_data,
    locations="iso_location_name",
    locationmode="country names",
    color="pca_score",
    hover_name="iso_location_name",
    animation_frame="year",
    color_continuous_scale="Turbo_r",
    title="Global QCI (PCA Score) - Age-standardized, Both Sexes"
)

# Update layout: increase size and font resolution
fig.update_layout(
    geo=dict(showframe=False, showcoastlines=False),
    width=1400,  # Increase width
    height=800,  # Increase height
    font=dict(size=16),  # Larger font
    title_font=dict(size=24),  # Larger title
    margin=dict(l=0, r=0, t=60, b=0)  # Tighter margin
)

fig.show()
```

```python
# Identify oldest and most recent years
most_recent_year = standardized_heatmap_data['year'].max()
oldest_year = standardized_heatmap_data['year'].min()
```

```python
# Get values for both years
latest_scores = standardized_heatmap_data[standardized_heatmap_data['year'] == most_recent_year][['iso_location_name', 'pca_score']]
latest_scores['year'] = most_recent_year

oldest_scores = standardized_heatmap_data[standardized_heatmap_data['year'] == oldest_year][['iso_location_name', 'pca_score']]
oldest_scores['year'] = oldest_year
```

```python
# Combine datasets
age_standard_heatmap_plot_data = pd.concat([oldest_scores, latest_scores])
```

```python
# Plot map
fig = px.choropleth(
    oldest_scores,
    locations="iso_location_name",
    locationmode="country names",
    color="pca_score",
    hover_name="iso_location_name",
    color_continuous_scale="Turbo_r",
    title=f"QCI Score - {oldest_year} (Age standardized, Both Sex)"
)
fig.update_layout(geo=dict(showframe=False, showcoastlines=False))
fig.show()
```

```python
# Plot map
fig = px.choropleth(
    latest_scores,
    locations="iso_location_name",
    locationmode="country names",
    color="pca_score",
    hover_name="iso_location_name",
    color_continuous_scale="Turbo_r",
    title=f"QCI Score - {most_recent_year} (Age standardized, Both Sex)"
)
fig.update_layout(geo=dict(showframe=False, showcoastlines=False))
fig.show()
```

```python
# Define SDI groups of interest
sdi_groups = ['Middle SDI', 'High-middle SDI', 'High SDI', 'Low SDI', 'Low-middle SDI']

# Filter the DataFrame
filtered_sdi_df = df[df['iso_location_name'].isin(sdi_groups)]
```

```python
filtered_sdi_df = filtered_sdi_df.rename(columns={'iso_location_name': 'sdi_group'})
```

```python
# Define SDI group colors 
color_map = {
    'Low SDI': '#d7191c',
    'Low-middle SDI': '#fdae61',
    'Middle SDI': '#ffffbf',
    'High-middle SDI': '#abdda4',
    'High SDI': '#2b83ba'
}
```

```python
# Sort SDI groups for plotting order
sdi_order = ['Low SDI', 'Low-middle SDI', 'Middle SDI', 'High-middle SDI', 'High SDI']
filtered_sdi_df['sdi_group'] = pd.Categorical(filtered_sdi_df['sdi_group'], categories=sdi_order, ordered=True)
```

```python
filtered_sdi_df = filtered_sdi_df[filtered_sdi_df["age_name"] == "Age-standardized"]
```

```python
filtered_sdi_df
```

```python
# Plot
plt.figure(figsize=(8, 10))
for group in sdi_order:
    subset = filtered_sdi_df[filtered_sdi_df['sdi_group'] == group]
    plt.scatter(
        [group] * len(subset),  # x: category repeated
        subset['qci'],          # y: QCI values
        color=color_map[group],
        edgecolor='black',
        linewidth=0.5,
        alpha=0.9,
        s=60,
        label=group
    )

plt.xlabel("Socio-Demographic Index Group (2021)", fontsize=12)
plt.ylabel("Quality of Care Index (QCI) 2021", fontsize=12)
plt.legend(title="SDI Group", loc="best")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
